# Import Library

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
# Set path to ChromeDriver
CHROMEDRIVER_PATH = "./chromedriver.exe"  # Update this path
CHROMEDRIVER_PATH_SERVER = "/usr/bin/chromedriver"

# Set Chrome options
options = Options()
options.add_argument("--start-maximized")  # Start Chrome maximized
options.add_argument("--disable-blink-features=AutomationControlled")  # Avoid detection
options.add_argument("--disable-infobars")  # Remove "Chrome is being controlled" banner
options.add_argument('--profile-directory=Default')
options.add_argument("--no-sandbox")  # Sometimes useful to avoid sandbox issues
options.add_argument("--disable-dev-shm-usage")  # Helps with some Docker environments
options.add_argument("--unsafely-treat-insecure-origin-as-secure=https://www.responsiblemineralsinitiative.org/facilities-lists/active-conformant-facilities-list/?")
# options.add_argument("--headless")

prefs = {
    "download.default_directory": r"C:\Users\Daniel_Lu\Documents\Projects_Jupyternotebook\CMRT_JupyterNotebook\Dynamic_List",  # Set your desired download directory
    "download.prompt_for_download": False,  # Disable download prompt
    "download.directory_upgrade": True,  # Auto-upgrade downloads
    "safebrowsing.enabled": True,  # Enable safe browsing
    "safebrowsing.disable_download_protection": True  # Disable security blocking
}
options.add_experimental_option("prefs", prefs)

# Set up WebDriver service
service = Service(CHROMEDRIVER_PATH_SERVER)
```

```python
# Initialize WebDriver
driver = webdriver.Chrome(service=service, options=options)

# Open a website
driver.get("https://www.responsiblemineralsinitiative.org/facilities-lists/active-conformant-facilities-list/")

# Print the page title
print(driver.title)

try:
    wait = WebDriverWait(driver, 10)
    accept_button = wait.until(EC.element_to_be_clickable((By.XPATH, "/html/body/div[5]/div[1]/div[2]/div/div/form/div/div[1]/p[2]/input")))
    #driver.execute_script("arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", accept_button)
    #driver.execute_script("arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", accept_button)
    driver.execute_script("window.scrollBy(0, 2000);")  # Scrolls 2000 pixels down

    
    # Locate the "I Accept" button using different strategies
    accept_button = driver.find_element(By.XPATH, "/html/body/div[5]/div[1]/div[2]/div/div/form/div/div[1]/p[2]/input")
    
    # Click the button
    accept_button.click()

    print("Clicked the 'I Accept' button successfully!")
    
except Exception as e:
    print(f"Error clicking the button: {e}")

wait = WebDriverWait(driver, 10)

driver.execute_script("window.scrollBy(0, 1000);")  # Scrolls 1000 pixels down
wait = WebDriverWait(driver, 10)

driver.switch_to.frame("Public Dynamic List")

download_data = driver.find_element(By.XPATH, "/html/body/div[1]/div/form/div/table[3]/tr/td/table/tbody/tr/td/div/a[2]")
download_data.click()
wait = WebDriverWait(driver, 15)
print("Clicked the 'Dwonload Data' button successfully!")

driver.implicitly_wait(10) 
download_excel = driver.find_element(By.XPATH, "/html/body/div[3]/div/div[1]")
download_excel.click()
driver.implicitly_wait(30) 
print("Download Excel successfully!")
# Close the browser after 5 seconds
import time
time.sleep(5)
driver.quit()
```
